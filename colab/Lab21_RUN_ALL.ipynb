```python
# NB6 §3 — chạy riêng phần hot-swap (merge_check.json đã có từ lần chạy trước)
import json, os, sys
os.chdir("/content/Day21-Track3-Finetuning-Lab"); sys.path.insert(0, "src")
from peft import PeftModel
from labkit import generate
from labkit.config import get_tier

print(open("results/merge_check.json").read())

target = [json.loads(l) for l in open("data/eval_target.jsonl", encoding="utf-8") if l.strip()]
model, tok = generate.load_base(get_tier("T4"))
model = PeftModel.from_pretrained(model, "adapters/correct", adapter_name="correct")
available = ["correct"]
for extra in ("attn_only", "qlora"):
    if os.path.exists(f"adapters/{extra}"):
        model.load_adapter(f"adapters/{extra}", adapter_name=extra)
        available.append(extra)

print("adapter đang nạp:", available)
for name in available:
    model.set_adapter(name)
    out, _ = generate.generate_batch(model, tok, [target[0]["input"]], system=generate.NAIVE_PROMPT)
    print(f"\n[{name}] -> {out[0][:140]}")
```
